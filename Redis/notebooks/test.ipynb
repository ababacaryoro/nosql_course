```python
!pip install redis
import redis
r = redis.Redis(host='redis', port=6379, db=0)
```

```python
# Set a key
r.set("user:99", "Alice")
```

```python
# Set multiple keys
r.mset({"user:98": "Bob", "user:32": "Charlie"})
```

```python
# Get a value
r.get("user:1")  # b'Alice'
```

```python
# Get multiple values
r.mget(["user:1", "user:2"])  # [b'Alice', b'Bob']
```

```python
# List all matching keys
r.keys("user:*")  
```

```python
# Overwrite value
r.set("user:99", "Alice Smith")
```

```python
# Increment value (e.g., score)
r.set("score", 100)
```

```python
r.incr("score")     # 101
```

```python
r.incrby("score", 5) # 106
```

```python
# Store and retrieve a dict.
r.hset('user-session:123', 
       mapping={
           'name': 'John',
           "surname": 'Smith',
           "company": 'Redis',
           "age": 29
       }
)
```

```python
# Retrieve
r.hgetall('user-session:123')
```

```python
## Retrive multiple values
r.scan(match = "user:*", count = 100)
```

```python
r.keys("user:1*")
```

```python
r.keys("*1*")
```

```python
r.close()
```
